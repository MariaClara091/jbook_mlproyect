```python
ax = sns.countplot(x='Target', data=dataset)
plt.title('Distribución de la variable Target')
plt.xlabel('Target')
plt.ylabel('Cantidad')

for p in ax.patches:
    ax.annotate(f'{int(p.get_height())}', 
                (p.get_x() + p.get_width() / 2, p.get_height()), 
                ha='center', va='bottom')

plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

# Calcular la matriz de correlación
dataset_num = dataset.select_dtypes(include=np.number)
corr = dataset_num.corr()
print("Matriz de correlación:", corr.shape)

# Crear la máscara
mask = np.zeros_like(corr)
mask[np.triu_indices_from(mask)] = True

# Configuración del mapa
plt.figure(figsize=(16, 12))  # Tamaño más razonable
sns.set(font_scale=0.8)       # Tamaño del texto general

sns.heatmap(
    corr, 
    mask=mask, 
    cmap='Reds', 
    annot=True, 
    fmt='.2f', 
    annot_kws={"size": 8},   # Tamaño del número dentro de la celda
    cbar=True
)

plt.xticks(rotation=45, ha="right")  # Rotar etiquetas X
plt.yticks(rotation=0)               # Mantener etiquetas Y horizontales
plt.title("Matriz de Correlación - Dataset de Spotify", fontsize=14)
plt.tight_layout()
plt.show()
```
